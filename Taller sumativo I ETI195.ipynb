**5.1 Análisis de variables**

Las variables para las cuales decidimos analizar sus distribuciones son las siguientes:

- Age (Edad) (Variable demográfica): El riesgo de desarrollar distintas enfermedades aumenta con la edad y la necesidad de adherirse a sistemas médicos, por lo cual nos parecio que esta variable tiene un impacto directo en la salud de las personas.
- Smoking (Fumar) (Variable sensible): De acuerdo a la BCRF existe una fuerte conexión entre fumar y el cancer de mamas. Estudios comprueban que mujeres fumadores tienen mayores probabilidades de desarrollar cancer que las no-fumadoras. Es importante conocer esta distribución.
- Annual_Income_USD (Ingresos anuales en dolares) (Variable demográfica): La posición economica tiene un impacto directo con nuestros habitos y y el acceso a salud, lo cual facilita o dificulta el desarrollo de enfermedades (dependiendo del nivel socioconomico).
  

```python
# Gráfico "Age"
print("--- Contabilización de Edad (Age) ---")

# Agrupamos la edad en intervalos (bins)
print(df["Age"].value_counts(bins=5).sort_index())

plt.figure(figsize=(8, 4))
sns.histplot(data=df, x='Age', kde=True)
plt.title("Distribución de la Edad")
plt.xlabel("Edad")
plt.ylabel("Cantidad de Pacientes")
plt.show()
```

```python
print("--- Contabilización de Hábito de Fumar ---")
print(df["Smoking"].value_counts())

plt.figure(figsize=(8, 4))
sns.countplot(data=df, x="Smoking")
plt.title("Distribución del Hábito de Fumar")
plt.xlabel("Fumador")
plt.ylabel("Cantidad de Pacientes")
plt.show()
```

```python
print("--- Contabilización de Ingresos Anuales ---")
# Agrupado en 5 intervalos para facilitar la lectura de los rangos de sueldo
print(df["Annual_Income_USD"].value_counts(bins=5).sort_index())

plt.figure(figsize=(8, 4))
sns.histplot(data=df, x="Annual_Income_USD", kde=True)
plt.title("Distribución de Ingresos Anuales")
plt.xlabel("Ingresos en Dólares (USD)")
plt.ylabel("Cantidad de Pacientes")
plt.show()
```
